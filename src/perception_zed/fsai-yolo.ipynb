```python
from ultralytics import YOLO

# Load a tiny pretrained model
model = YOLO("/kaggle/input/yolo11n-1oepochs/yolov8s_50epochs.pt")

# Train
results = model.train(
    data="/kaggle/input/fsoco-split/data_split/config.yaml",  
    imgsz=640,
    epochs=20,      
    batch=10,
    name="FSOCO_YOLOv8s_80pochs",
    project="/kaggle/working",
    exist_ok=True,
    val=True,
    device='cuda'  
)
```

```python
test_results = model.val(
    data="/kaggle/input/fsoco-split/data_split/config.yaml",
    split="test",
    batch=10,
    imgsz=640,
    save=True          # save test predictions
)
```
